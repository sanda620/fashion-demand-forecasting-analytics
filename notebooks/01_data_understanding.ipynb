```python
import numpy as np
import pandas as pd

sales = pd.read_csv("../data/raw/sales.csv")
train = pd.read_csv("../data/raw/stfore_train.csv")
test = pd.read_csv("../data/raw/stfore_test.csv")

# Inspect normalization scalar
scalar = np.load("../data/raw/stfore_sales_norm_scalar.npy")

print("Normalization scalar:", scalar)

print("\nDataset shapes:")
print("Sales:", sales.shape)
print("Training:", train.shape)
print("Testing:", test.shape)

# Inspect weekly sales columns
weekly_cols = [str(i) for i in range(12)]

print("\nOriginal sales values:")
display(sales[weekly_cols].head())

print("\nNormalized training values:")
display(train[weekly_cols].head())

print("\nNormalized test values:")
display(test[weekly_cols].head())

# Check whether sales values are normalized
print("\nOriginal first row:")
print(sales.loc[0, weekly_cols].values)

print("\nNormalized first row:")
print(train.loc[0, weekly_cols].values)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

key_cols = ["external_code", "retail", "season", "release_date"]
weekly_cols = [str(i) for i in range(12)]
value_cols = ["restock"] + weekly_cols

# Combine the supplied training and testing datasets
combined = pd.concat([train, test], ignore_index=True)

print("Duplicate keys:")
print("Sales:", sales.duplicated(key_cols).sum())
print("Combined:", combined.duplicated(key_cols).sum())

print("\nTrain-test overlap:")
train_keys = set(map(tuple, train[key_cols].to_numpy()))
test_keys = set(map(tuple, test[key_cols].to_numpy()))
print("Overlapping records:", len(train_keys & test_keys))

# Verify normalized values against original sales
if (sales.duplicated(key_cols).sum() == 0
        and combined.duplicated(key_cols).sum() == 0):

    comparison = sales[key_cols + value_cols].merge(
        combined[key_cols + value_cols],
        on=key_cols,
        how="inner",
        suffixes=("_raw", "_normalized"),
        validate="one_to_one"
    )

    print("\nMatched records:", len(comparison))

    raw_values = comparison[
        [f"{c}_raw" for c in value_cols]
    ].to_numpy()

    normalized_values = comparison[
        [f"{c}_normalized" for c in value_cols]
    ].to_numpy()

    print(
        "Maximum normalization difference:",
        np.max(np.abs(raw_values - 53 * normalized_values))
    )
else:
    print("\nDuplicate keys found. We need to investigate before matching.")
```
